```python
# ======================================================
# HOUSE PRICE DATA PREPROCESSING PROJECT
# ======================================================

# ===========================
# 1. Import Libraries
# ===========================

import pandas as pd
import sqlite3
from google.colab import files

print("Libraries Imported Successfully\n")

# ===========================
# 2. Upload Dataset
# ===========================

print("Upload your 'housing prices.csv' file")
uploaded = files.upload()

# ===========================
# 3. Load Dataset
# ===========================

df = pd.read_csv("housing prices.csv")

print("\nDataset Loaded Successfully!")

# ===========================
# 4. Dataset Information
# ===========================

print("\n==============================")
print("FIRST 5 ROWS")
print("==============================")
display(df.head())

print("\n==============================")
print("LAST 5 ROWS")
print("==============================")
display(df.tail())

print("\nShape of Dataset")
print(df.shape)

print("\nColumns")
print(df.columns.tolist())

print("\nData Types")
print(df.dtypes)

print("\nDataset Information")
df.info()

# ===========================
# 5. Data Cleaning
# ===========================

print("\nCleaning Column Names...")

df.columns = df.columns.str.strip()

print("Column Names After Cleaning")
print(df.columns.tolist())

# ===========================
# 6. Missing Values
# ===========================

print("\n==============================")
print("MISSING VALUES")
print("==============================")

print(df.isnull().sum())

# Fill numerical missing values

df["total_bedrooms"] = df["total_bedrooms"].fillna(
    df["total_bedrooms"].median()
)

print("\nMissing Values After Filling")
print(df.isnull().sum())

# ===========================
# 7. Remove Duplicates
# ===========================

print("\n==============================")
print("DUPLICATE VALUES")
print("==============================")

duplicates = df.duplicated().sum()

print("Duplicate Rows :", duplicates)

df = df.drop_duplicates()

print("Duplicates Removed Successfully!")

print("New Shape :", df.shape)

# ===========================
# 8. Statistical Summary
# ===========================

print("\n==============================")
print("STATISTICAL SUMMARY")
print("==============================")

display(df.describe())

print("\nCategorical Summary")

display(df.describe(include="object"))

# ===========================
# 9. Export CSV
# ===========================

df.to_csv("cleaned_housing_prices.csv", index=False)

print("\nCSV File Created!")

# ===========================
# 10. Export Excel
# ===========================

df.to_excel("cleaned_housing_prices.xlsx", index=False)

print("Excel File Created!")

# ===========================
# 11. Export JSON
# ===========================

df.to_json(
    "cleaned_housing_prices.json",
    orient="records",
    indent=4
)

print("JSON File Created!")

# ===========================
# 12. Export SQL
# ===========================

connection = sqlite3.connect("housing.db")

df.to_sql(
    "housing",
    connection,
    if_exists="replace",
    index=False
)

print("SQLite Database Created!")

# ===========================
# 13. SQL Queries
# ===========================

print("\n==============================")
print("SELECT")
print("==============================")

query = "SELECT * FROM housing LIMIT 5"

display(pd.read_sql(query, connection))

print("\n==============================")
print("WHERE")
print("==============================")

query = """
SELECT *
FROM housing
WHERE median_income > 5
LIMIT 5
"""

display(pd.read_sql(query, connection))

print("\n==============================")
print("ORDER BY")
print("==============================")

query = """
SELECT *
FROM housing
ORDER BY median_house_value DESC
LIMIT 5
"""

display(pd.read_sql(query, connection))

print("\n==============================")
print("GROUP BY")
print("==============================")

query = """
SELECT
ocean_proximity,
AVG(median_house_value) AS AveragePrice
FROM housing
GROUP BY ocean_proximity
"""

display(pd.read_sql(query, connection))

print("\n==============================")
print("COUNT")
print("==============================")

query = """
SELECT COUNT(*) AS TotalRows
FROM housing
"""

display(pd.read_sql(query, connection))

print("\n==============================")
print("AVG")
print("==============================")

query = """
SELECT AVG(median_income) AS AverageIncome
FROM housing
"""

display(pd.read_sql(query, connection))

print("\n==============================")
print("MAX")
print("==============================")

query = """
SELECT MAX(median_house_value) AS HighestPrice
FROM housing
"""

display(pd.read_sql(query, connection))

print("\n==============================")
print("MIN")
print("==============================")

query = """
SELECT MIN(median_house_value) AS LowestPrice
FROM housing
"""

display(pd.read_sql(query, connection))

# ===========================
# 14. Custom SQL Query
# ===========================

print("\n==============================")
print("CUSTOM SQL QUERY")
print("==============================")

query = """
SELECT
ocean_proximity,
COUNT(*) AS NumberOfHouses,
AVG(median_house_value) AS AveragePrice
FROM housing
GROUP BY ocean_proximity
ORDER BY AveragePrice DESC
"""

result = pd.read_sql(query, connection)

display(result)

connection.close()

# ===========================
# 15. Download Files
# ===========================

print("\nDownloading Files...")

files.download("cleaned_housing_prices.csv")
files.download("cleaned_housing_prices.xlsx")
files.download("cleaned_housing_prices.json")
files.download("housing.db")

print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
