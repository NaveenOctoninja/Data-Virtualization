```python
# ============================================================
# EXPERIMENT 5
# California Housing Price Analysis and Prediction
# DATA VISUALIZATION
# ============================================================


# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns


# Set visualization style
sns.set_style("whitegrid")

plt.rcParams["figure.figsize"] = (10, 6)


# ============================================================
# 2. LOAD THE DATASET
# ============================================================

# Change the filename if required

df = pd.read_csv("/content/housing prices.csv")

print("Dataset Loaded Successfully!")

print("\nFirst 5 Rows:")
print(df.head())


# ============================================================
# 3. BASIC DATASET INFORMATION
# ============================================================

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()


# ============================================================
# 4. HANDLE MISSING VALUES
# ============================================================

# Create a copy for visualization

df_visual = df.copy()


# Fill numerical missing values using median

numeric_columns = df_visual.select_dtypes(
    include=np.number
).columns


for column in numeric_columns:

    df_visual[column] = df_visual[column].fillna(
        df_visual[column].median()
    )


# Fill categorical missing values using mode

categorical_columns = df_visual.select_dtypes(
    include="object"
).columns


for column in categorical_columns:

    df_visual[column] = df_visual[column].fillna(
        df_visual[column].mode()[0]
    )


print("\nMissing Values After Handling:")

print(
    df_visual.isnull().sum()
)


# ============================================================
# 5. HISTOGRAM - MEDIAN HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df_visual,
    x="median_house_value",
    bins=30,
    kde=True
)

plt.title(
    "Distribution of Median House Value"
)

plt.xlabel(
    "Median House Value"
)

plt.ylabel(
    "Frequency"
)

plt.show()


# ============================================================
# 6. HISTOGRAM - MEDIAN INCOME
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df_visual,
    x="median_income",
    bins=30,
    kde=True
)

plt.title(
    "Distribution of Median Income"
)

plt.xlabel(
    "Median Income"
)

plt.ylabel(
    "Frequency"
)

plt.show()


# ============================================================
# 7. HISTOGRAM - HOUSING MEDIAN AGE
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    data=df_visual,
    x="housing_median_age",
    bins=30,
    kde=True
)

plt.title(
    "Distribution of Housing Median Age"
)

plt.xlabel(
    "Housing Median Age"
)

plt.ylabel(
    "Frequency"
)

plt.show()


# ============================================================
# 8. COUNT PLOT - OCEAN PROXIMITY
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df_visual,
    x="ocean_proximity"
)

plt.title(
    "Number of Housing Districts by Ocean Proximity"
)

plt.xlabel(
    "Ocean Proximity"
)

plt.ylabel(
    "Number of Districts"
)

plt.xticks(
    rotation=45
)

plt.show()


# ============================================================
# 9. SCATTER PLOT - MEDIAN INCOME VS HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_visual,
    x="median_income",
    y="median_house_value",
    alpha=0.4
)

plt.title(
    "Median Income vs Median House Value"
)

plt.xlabel(
    "Median Income"
)

plt.ylabel(
    "Median House Value"
)

plt.show()


# ============================================================
# 10. SCATTER PLOT - TOTAL ROOMS VS HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_visual,
    x="total_rooms",
    y="median_house_value",
    alpha=0.4
)

plt.title(
    "Total Rooms vs Median House Value"
)

plt.xlabel(
    "Total Rooms"
)

plt.ylabel(
    "Median House Value"
)

plt.show()


# ============================================================
# 11. SCATTER PLOT - POPULATION VS HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_visual,
    x="population",
    y="median_house_value",
    alpha=0.4
)

plt.title(
    "Population vs Median House Value"
)

plt.xlabel(
    "Population"
)

plt.ylabel(
    "Median House Value"
)

plt.show()


# ============================================================
# 12. SCATTER PLOT - HOUSING AGE VS HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_visual,
    x="housing_median_age",
    y="median_house_value",
    alpha=0.4
)

plt.title(
    "Housing Median Age vs Median House Value"
)

plt.xlabel(
    "Housing Median Age"
)

plt.ylabel(
    "Median House Value"
)

plt.show()


# ============================================================
# 13. BOX PLOT - MEDIAN HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df_visual,
    x="median_house_value"
)

plt.title(
    "Box Plot of Median House Value"
)

plt.xlabel(
    "Median House Value"
)

plt.show()


# ============================================================
# 14. BOX PLOT - MEDIAN INCOME
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df_visual,
    x="median_income"
)

plt.title(
    "Box Plot of Median Income"
)

plt.xlabel(
    "Median Income"
)

plt.show()


# ============================================================
# 15. BOX PLOT - HOUSE VALUE BY OCEAN PROXIMITY
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df_visual,
    x="ocean_proximity",
    y="median_house_value"
)

plt.title(
    "Median House Value by Ocean Proximity"
)

plt.xlabel(
    "Ocean Proximity"
)

plt.ylabel(
    "Median House Value"
)

plt.xticks(
    rotation=45
)

plt.show()


# ============================================================
# 16. VIOLIN PLOT - HOUSE VALUE BY OCEAN PROXIMITY
# ============================================================

plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df_visual,
    x="ocean_proximity",
    y="median_house_value"
)

plt.title(
    "Distribution of House Value by Ocean Proximity"
)

plt.xlabel(
    "Ocean Proximity"
)

plt.ylabel(
    "Median House Value"
)

plt.xticks(
    rotation=45
)

plt.show()


# ============================================================
# 17. BAR CHART - AVERAGE HOUSE VALUE
# ============================================================

average_house_value = (
    df_visual
    .groupby("ocean_proximity")
    ["median_house_value"]
    .mean()
    .sort_values(
        ascending=False
    )
)


plt.figure(figsize=(10, 6))

average_house_value.plot(
    kind="bar"
)

plt.title(
    "Average House Value by Ocean Proximity"
)

plt.xlabel(
    "Ocean Proximity"
)

plt.ylabel(
    "Average House Value"
)

plt.xticks(
    rotation=45
)

plt.show()


# ============================================================
# 18. BAR CHART - AVERAGE MEDIAN INCOME
# ============================================================

average_income = (
    df_visual
    .groupby("ocean_proximity")
    ["median_income"]
    .mean()
    .sort_values(
        ascending=False
    )
)


plt.figure(figsize=(10, 6))

average_income.plot(
    kind="bar"
)

plt.title(
    "Average Median Income by Ocean Proximity"
)

plt.xlabel(
    "Ocean Proximity"
)

plt.ylabel(
    "Average Median Income"
)

plt.xticks(
    rotation=45
)

plt.show()


# ============================================================
# 19. CORRELATION HEATMAP
# ============================================================

numeric_df = df_visual.select_dtypes(
    include=np.number
)


correlation = numeric_df.corr()


plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title(
    "Correlation Heatmap - California Housing Dataset"
)

plt.show()


# ============================================================
# 20. CORRELATION WITH HOUSE VALUE
# ============================================================

house_value_corr = (
    correlation[
        "median_house_value"
    ]
    .sort_values(
        ascending=False
    )
)


print("\nCorrelation with Median House Value:")

print(
    house_value_corr
)


# ============================================================
# 21. GEOGRAPHICAL SCATTER PLOT
# ============================================================

plt.figure(figsize=(10, 7))

plt.scatter(
    df_visual["longitude"],
    df_visual["latitude"],
    alpha=0.3,
    s=10
)

plt.title(
    "Geographical Distribution of California Housing"
)

plt.xlabel(
    "Longitude"
)

plt.ylabel(
    "Latitude"
)

plt.show()


# ============================================================
# 22. GEOGRAPHICAL PLOT COLOURED BY HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    df_visual["longitude"],
    df_visual["latitude"],
    c=df_visual["median_house_value"],
    cmap="viridis",
    alpha=0.4,
    s=10
)


plt.colorbar(
    scatter,
    label="Median House Value"
)

plt.title(
    "California Housing Prices by Geographic Location"
)

plt.xlabel(
    "Longitude"
)

plt.ylabel(
    "Latitude"
)

plt.show()


# ============================================================
# 23. GEOGRAPHICAL PLOT WITH POPULATION
# ============================================================

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    df_visual["longitude"],
    df_visual["latitude"],
    s=df_visual["population"] / 50,
    c=df_visual["median_house_value"],
    cmap="viridis",
    alpha=0.4
)


plt.colorbar(
    scatter,
    label="Median House Value"
)

plt.title(
    "California Housing: Location, Population and House Value"
)

plt.xlabel(
    "Longitude"
)

plt.ylabel(
    "Latitude"
)

plt.show()


# ============================================================
# 24. PAIRPLOT
# ============================================================

pairplot_columns = [
    "median_house_value",
    "median_income",
    "housing_median_age",
    "total_rooms",
    "population"
]


sns.pairplot(
    df_visual[
        pairplot_columns
    ]
)

plt.show()


# ============================================================
# 25. BUBBLE CHART
# ============================================================

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    df_visual["median_income"],
    df_visual["median_house_value"],

    # Population determines bubble size
    s=df_visual["population"] / 20,

    # Housing age determines bubble colour
    c=df_visual["housing_median_age"],

    cmap="viridis",

    alpha=0.5
)


plt.colorbar(
    scatter,
    label="Housing Median Age"
)

plt.title(
    "Bubble Chart: Income vs House Value"
)

plt.xlabel(
    "Median Income"
)

plt.ylabel(
    "Median House Value"
)

plt.show()


# ============================================================
# 26. HOUSE PRICE CATEGORIES
# ============================================================

df_visual["price_category"] = pd.cut(
    df_visual["median_house_value"],

    bins=[
        0,
        100000,
        200000,
        300000,
        400000,
        500000,
        np.inf
    ],

    labels=[
        "Very Low",
        "Low",
        "Medium",
        "High",
        "Very High",
        "Premium"
    ]
)


print(
    "\nHouse Price Categories:"
)

print(
    df_visual[
        "price_category"
    ].value_counts()
)


# ============================================================
# 27. HOUSE PRICE CATEGORY COUNT PLOT
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df_visual,
    x="price_category"
)

plt.title(
    "Distribution of House Price Categories"
)

plt.xlabel(
    "Price Category"
)

plt.ylabel(
    "Number of Houses"
)

plt.xticks(
    rotation=45
)

plt.show()


# ============================================================
# 28. PRICE CATEGORY BY OCEAN PROXIMITY
# ============================================================

plt.figure(figsize=(12, 6))

sns.countplot(
    data=df_visual,
    x="ocean_proximity",
    hue="price_category"
)

plt.title(
    "House Price Categories by Ocean Proximity"
)

plt.xlabel(
    "Ocean Proximity"
)

plt.ylabel(
    "Number of Houses"
)

plt.xticks(
    rotation=45
)

plt.show()


# ============================================================
# 29. HEATMAP OF PRICE CATEGORY VS OCEAN PROXIMITY
# ============================================================

price_ocean = pd.crosstab(
    df_visual["ocean_proximity"],
    df_visual["price_category"]
)


plt.figure(figsize=(10, 6))

sns.heatmap(
    price_ocean,
    annot=True,
    fmt="d",
    cmap="YlOrRd"
)

plt.title(
    "House Price Category vs Ocean Proximity"
)

plt.xlabel(
    "Price Category"
)

plt.ylabel(
    "Ocean Proximity"
)

plt.show()


# ============================================================
# 30. FINAL VISUALIZATION SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("EXPERIMENT 5 VISUALIZATION SUMMARY")
print("=" * 60)

print("""
The following visualization techniques were performed:

1. Histogram
2. KDE Distribution Plot
3. Count Plot
4. Scatter Plot
5. Box Plot
6. Violin Plot
7. Bar Chart
8. Correlation Heatmap
9. Geographic Scatter Plot
10. Population Geographic Plot
11. Pair Plot
12. Bubble Chart
13. Price Category Visualization
14. Category Comparison Plot
15. Cross-tabulation Heatmap
""")


# ============================================================
# 31. EXPERIMENT 5 COMPLETED
# ============================================================

print("\n")
print("=" * 60)
print("EXPERIMENT 5 COMPLETED SUCCESSFULLY!")
print("=" * 60)
```
