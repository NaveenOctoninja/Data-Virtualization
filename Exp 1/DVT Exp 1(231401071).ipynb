```python
import pandas as pd
import numpy as np

# Load dataset
path = "/content/housing prices.csv"
df = pd.read_csv(path)

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nDataset information:")
df.info()

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nStatistical description:")
display(df.describe())

print("\nMissing values:")
display(df.isnull().sum())

# Data cleaning
df = df.dropna(how="all")
df = df.drop_duplicates()

df["ocean_proximity"] = (
    df["ocean_proximity"].astype(str).str.strip().str.upper()
)

numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())

print("\nMissing values after cleaning:")
display(df.isnull().sum())

print("\nCleaned dataset shape:")
print(df.shape)

# Housing value analysis
print("\nHousing Value Statistics:")
display(
    df["median_house_value"].agg(
        ["count", "mean", "median", "std", "min", "max"]
    )
)

# Ocean proximity analysis
print("\nOcean-Proximity-wise Housing Analysis:")
display(
    df.groupby("ocean_proximity")["median_house_value"]
      .agg(["count", "mean"])
      .sort_values("mean", ascending=False)
)

# Feature-wise analysis
print("\nHousing Feature-wise Statistics:")
display(
    df[
        [
            "total_rooms",
            "total_bedrooms",
            "population",
            "households",
            "median_income",
            "median_house_value"
        ]
    ].describe()
)

# Correlation
print("\nCorrelation:")
display(df.corr(numeric_only=True))

# Top 10 houses
print("\nTop 10 Houses by Median House Value:")
display(
    df.sort_values("median_house_value", ascending=False)
      .head(10)[
          [
              "longitude",
              "latitude",
              "median_income",
              "total_rooms",
              "median_house_value",
              "ocean_proximity"
          ]
      ]
)

# Top 10 by income
print("\nTop 10 Records by Median Income:")
display(
    df.sort_values("median_income", ascending=False)
      .head(10)[
          [
              "longitude",
              "latitude",
              "median_income",
              "median_house_value",
              "ocean_proximity"
          ]
      ]
)

# Export cleaned dataset
df.to_csv("housing_prices_cleaned.csv", index=False)
df.to_excel("housing_prices_cleaned.xlsx", index=False)
df.to_json("housing_prices_cleaned.json", orient="records", indent=4)

print("\nCSV, Excel and JSON files created successfully.")
```
