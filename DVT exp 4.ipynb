```python
# ============================================================
# EXPERIMENT 4
# California Housing Price Analysis and Prediction
# DATA ANALYSIS
# ============================================================


# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 2. LOAD THE DATASET
# ============================================================

# Change the filename if required

df = pd.read_csv("/content/housing prices.csv")

print("Dataset Loaded Successfully!")

print("\nFirst 5 Rows:")
print(df.head())


# ============================================================
# 3. DATASET INFORMATION
# ============================================================

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()


# ============================================================
# 4. DESCRIPTIVE STATISTICS
# ============================================================

print("\nDescriptive Statistics:")
print(df.describe())


# ============================================================
# 5. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# 6. HANDLE MISSING VALUES
# ============================================================

# Create a copy for analysis

df_analysis = df.copy()


# Fill numerical missing values with median

numeric_columns = df_analysis.select_dtypes(
    include=np.number
).columns


for column in numeric_columns:

    df_analysis[column] = df_analysis[column].fillna(
        df_analysis[column].median()
    )


# Fill categorical missing values with mode

categorical_columns = df_analysis.select_dtypes(
    include="object"
).columns


for column in categorical_columns:

    df_analysis[column] = df_analysis[column].fillna(
        df_analysis[column].mode()[0]
    )


print("\nMissing Values After Handling:")
print(df_analysis.isnull().sum())


# ============================================================
# 7. FIND UNIQUE OCEAN PROXIMITY VALUES
# ============================================================

print("\nUnique Ocean Proximity Categories:")

print(
    df_analysis["ocean_proximity"].unique()
)


# ============================================================
# 8. COUNT HOUSES BY OCEAN PROXIMITY
# ============================================================

print("\nNumber of Houses by Ocean Proximity:")

print(
    df_analysis[
        "ocean_proximity"
    ].value_counts()
)


# ============================================================
# 9. FILTER HIGH-VALUE HOUSES
# ============================================================

print("\n")
print("=" * 60)
print("FILTERING HIGH-VALUE HOUSES")
print("=" * 60)


high_value_houses = df_analysis[
    df_analysis["median_house_value"] > 400000
]


print(
    "Number of Houses with Value Greater Than $400,000:"
)

print(
    len(high_value_houses)
)


print("\nFirst 10 High-Value Houses:")

print(
    high_value_houses.head(10)
)


# ============================================================
# 10. FILTER LOW-INCOME AREAS
# ============================================================

low_income_areas = df_analysis[
    df_analysis["median_income"] < 3
]


print("\nNumber of Low-Income Areas:")

print(
    len(low_income_areas)
)


# ============================================================
# 11. FILTER HIGH-INCOME AREAS
# ============================================================

high_income_areas = df_analysis[
    df_analysis["median_income"] > 8
]


print("\nNumber of High-Income Areas:")

print(
    len(high_income_areas)
)


# ============================================================
# 12. FILTER HOUSES NEAR BAY
# ============================================================

near_bay = df_analysis[
    df_analysis["ocean_proximity"] == "NEAR BAY"
]


print("\nNumber of Houses Near Bay:")

print(
    len(near_bay)
)


print("\nFirst 10 Houses Near Bay:")

print(
    near_bay.head(10)
)


# ============================================================
# 13. FILTER ISLAND HOUSES
# ============================================================

island_houses = df_analysis[
    df_analysis["ocean_proximity"] == "ISLAND"
]


print("\nNumber of Island Houses:")

print(
    len(island_houses)
)


# ============================================================
# 14. SORT BY MEDIAN HOUSE VALUE
# ============================================================

print("\n")
print("=" * 60)
print("SORTING DATA")
print("=" * 60)


sorted_by_value = df_analysis.sort_values(
    by="median_house_value",
    ascending=False
)


print("\nTop 10 Most Expensive Housing Districts:")

print(
    sorted_by_value.head(10)
)


# ============================================================
# 15. SORT BY MEDIAN INCOME
# ============================================================

sorted_by_income = df_analysis.sort_values(
    by="median_income",
    ascending=False
)


print("\nTop 10 Highest Income Housing Districts:")

print(
    sorted_by_income.head(10)
)


# ============================================================
# 16. SORT BY POPULATION
# ============================================================

sorted_by_population = df_analysis.sort_values(
    by="population",
    ascending=False
)


print("\nTop 10 Most Populated Housing Districts:")

print(
    sorted_by_population.head(10)
)


# ============================================================
# 17. MEAN OF MEDIAN HOUSE VALUE
# ============================================================

mean_house_value = (
    df_analysis["median_house_value"].mean()
)


print("\nMean Median House Value:")

print(
    mean_house_value
)


# ============================================================
# 18. MEDIAN OF MEDIAN HOUSE VALUE
# ============================================================

median_house_value = (
    df_analysis["median_house_value"].median()
)


print("\nMedian Median House Value:")

print(
    median_house_value
)


# ============================================================
# 19. MODE OF MEDIAN HOUSE VALUE
# ============================================================

mode_house_value = (
    df_analysis["median_house_value"].mode()
)


print("\nMode of Median House Value:")

print(
    mode_house_value
)


# ============================================================
# 20. STANDARD DEVIATION
# ============================================================

std_house_value = (
    df_analysis["median_house_value"].std()
)


print("\nStandard Deviation of House Value:")

print(
    std_house_value
)


# ============================================================
# 21. MINIMUM AND MAXIMUM HOUSE VALUE
# ============================================================

print("\nMinimum House Value:")

print(
    df_analysis["median_house_value"].min()
)


print("\nMaximum House Value:")

print(
    df_analysis["median_house_value"].max()
)


# ============================================================
# 22. RANGE OF HOUSE VALUES
# ============================================================

house_value_range = (
    df_analysis["median_house_value"].max()
    -
    df_analysis["median_house_value"].min()
)


print("\nRange of House Values:")

print(
    house_value_range
)


# ============================================================
# 23. AVERAGE HOUSE VALUE BY OCEAN PROXIMITY
# ============================================================

print("\n")
print("=" * 60)
print("GROUPING AND AGGREGATION")
print("=" * 60)


average_value_by_ocean = (
    df_analysis
    .groupby("ocean_proximity")
    ["median_house_value"]
    .mean()
    .sort_values(
        ascending=False
    )
)


print("\nAverage House Value by Ocean Proximity:")

print(
    average_value_by_ocean
)


# ============================================================
# 24. MEDIAN HOUSE VALUE BY OCEAN PROXIMITY
# ============================================================

median_value_by_ocean = (
    df_analysis
    .groupby("ocean_proximity")
    ["median_house_value"]
    .median()
)


print("\nMedian House Value by Ocean Proximity:")

print(
    median_value_by_ocean
)


# ============================================================
# 25. AVERAGE MEDIAN INCOME BY OCEAN PROXIMITY
# ============================================================

average_income_by_ocean = (
    df_analysis
    .groupby("ocean_proximity")
    ["median_income"]
    .mean()
    .sort_values(
        ascending=False
    )
)


print("\nAverage Median Income by Ocean Proximity:")

print(
    average_income_by_ocean
)


# ============================================================
# 26. AVERAGE POPULATION BY OCEAN PROXIMITY
# ============================================================

average_population_by_ocean = (
    df_analysis
    .groupby("ocean_proximity")
    ["population"]
    .mean()
)


print("\nAverage Population by Ocean Proximity:")

print(
    average_population_by_ocean
)


# ============================================================
# 27. MULTIPLE AGGREGATION
# ============================================================

group_analysis = (
    df_analysis
    .groupby("ocean_proximity")
    .agg(
        {
            "median_house_value": [
                "mean",
                "median",
                "min",
                "max"
            ],

            "median_income": [
                "mean",
                "median"
            ],

            "population": [
                "mean",
                "sum"
            ]
        }
    )
)


print("\nMultiple Aggregation by Ocean Proximity:")

print(
    group_analysis
)


# ============================================================
# 28. CORRELATION ANALYSIS
# ============================================================

print("\n")
print("=" * 60)
print("CORRELATION ANALYSIS")
print("=" * 60)


correlation = df_analysis.corr(
    numeric_only=True
)


print("\nCorrelation Matrix:")

print(
    correlation
)


# ============================================================
# 29. CORRELATION WITH HOUSE VALUE
# ============================================================

house_value_correlation = (
    correlation[
        "median_house_value"
    ]
    .sort_values(
        ascending=False
    )
)


print(
    "\nCorrelation with Median House Value:"
)

print(
    house_value_correlation
)


# ============================================================
# 30. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title(
    "Correlation Heatmap - California Housing"
)

plt.show()


# ============================================================
# 31. MEDIAN INCOME VS HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_analysis,
    x="median_income",
    y="median_house_value",
    alpha=0.4
)

plt.title(
    "Median Income vs Median House Value"
)

plt.xlabel(
    "Median Income"
)

plt.ylabel(
    "Median House Value"
)

plt.show()


# ============================================================
# 32. HOUSING AGE VS HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_analysis,
    x="housing_median_age",
    y="median_house_value",
    alpha=0.4
)

plt.title(
    "Housing Median Age vs Median House Value"
)

plt.xlabel(
    "Housing Median Age"
)

plt.ylabel(
    "Median House Value"
)

plt.show()


# ============================================================
# 33. TOTAL ROOMS VS HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_analysis,
    x="total_rooms",
    y="median_house_value",
    alpha=0.4
)

plt.title(
    "Total Rooms vs Median House Value"
)

plt.xlabel(
    "Total Rooms"
)

plt.ylabel(
    "Median House Value"
)

plt.show()


# ============================================================
# 34. POPULATION VS HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_analysis,
    x="population",
    y="median_house_value",
    alpha=0.4
)

plt.title(
    "Population vs Median House Value"
)

plt.xlabel(
    "Population"
)

plt.ylabel(
    "Median House Value"
)

plt.show()


# ============================================================
# 35. BAR CHART - AVERAGE HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 6))

average_value_by_ocean.plot(
    kind="bar"
)

plt.title(
    "Average House Value by Ocean Proximity"
)

plt.xlabel(
    "Ocean Proximity"
)

plt.ylabel(
    "Average House Value"
)

plt.xticks(
    rotation=45
)

plt.show()


# ============================================================
# 36. BAR CHART - AVERAGE INCOME
# ============================================================

plt.figure(figsize=(10, 6))

average_income_by_ocean.plot(
    kind="bar"
)

plt.title(
    "Average Median Income by Ocean Proximity"
)

plt.xlabel(
    "Ocean Proximity"
)

plt.ylabel(
    "Average Median Income"
)

plt.xticks(
    rotation=45
)

plt.show()


# ============================================================
# 37. BOX PLOT BY OCEAN PROXIMITY
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df_analysis,
    x="ocean_proximity",
    y="median_house_value"
)

plt.title(
    "House Value Distribution by Ocean Proximity"
)

plt.xlabel(
    "Ocean Proximity"
)

plt.ylabel(
    "Median House Value"
)

plt.xticks(
    rotation=45
)

plt.show()


# ============================================================
# 38. GEOGRAPHICAL ANALYSIS
# ============================================================

plt.figure(figsize=(10, 7))

plt.scatter(
    df_analysis["longitude"],
    df_analysis["latitude"],
    alpha=0.3
)

plt.title(
    "Geographical Distribution of California Housing"
)

plt.xlabel(
    "Longitude"
)

plt.ylabel(
    "Latitude"
)

plt.show()


# ============================================================
# 39. GEOGRAPHICAL ANALYSIS BY HOUSE VALUE
# ============================================================

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    df_analysis["longitude"],
    df_analysis["latitude"],
    c=df_analysis["median_house_value"],
    cmap="viridis",
    alpha=0.4,
    s=10
)

plt.colorbar(
    scatter,
    label="Median House Value"
)

plt.title(
    "Geographical Distribution by House Value"
)

plt.xlabel(
    "Longitude"
)

plt.ylabel(
    "Latitude"
)

plt.show()


# ============================================================
# 40. CREATE HOUSE PRICE CATEGORIES
# ============================================================

print("\n")
print("=" * 60)
print("HOUSE PRICE CATEGORIZATION")
print("=" * 60)


df_analysis["price_category"] = pd.cut(
    df_analysis["median_house_value"],

    bins=[
        0,
        100000,
        200000,
        300000,
        400000,
        500000,
        np.inf
    ],

    labels=[
        "Very Low",
        "Low",
        "Medium",
        "High",
        "Very High",
        "Premium"
    ]
)


print(
    "\nHouse Price Category Counts:"
)

print(
    df_analysis[
        "price_category"
    ].value_counts()
)


# ============================================================
# 41. PRICE CATEGORY BY OCEAN PROXIMITY
# ============================================================

price_ocean = pd.crosstab(
    df_analysis["ocean_proximity"],
    df_analysis["price_category"]
)


print(
    "\nHouse Price Category by Ocean Proximity:"
)

print(
    price_ocean
)


# ============================================================
# 42. VISUALIZE PRICE CATEGORIES
# ============================================================

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df_analysis,
    x="price_category"
)

plt.title(
    "Distribution of House Price Categories"
)

plt.xlabel(
    "Price Category"
)

plt.ylabel(
    "Number of Houses"
)

plt.xticks(
    rotation=45
)

plt.show()


# ============================================================
# 43. FIND HIGHEST VALUE HOUSING AREA
# ============================================================

highest_value_area = (
    average_value_by_ocean.idxmax()
)

highest_value = (
    average_value_by_ocean.max()
)


print(
    "\nOcean Proximity with Highest Average House Value:"
)

print(
    highest_value_area
)

print(
    "\nHighest Average House Value:"
)

print(
    highest_value
)


# ============================================================
# 44. FIND LOWEST VALUE HOUSING AREA
# ============================================================

lowest_value_area = (
    average_value_by_ocean.idxmin()
)

lowest_value = (
    average_value_by_ocean.min()
)


print(
    "\nOcean Proximity with Lowest Average House Value:"
)

print(
    lowest_value_area
)

print(
    "\nLowest Average House Value:"
)

print(
    lowest_value
)


# ============================================================
# 45. FINAL ANALYSIS SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("EXPERIMENT 4 ANALYSIS SUMMARY")
print("=" * 60)


print("""
The following data analysis operations were performed:

1. Dataset exploration
2. Missing value handling
3. Filtering
4. Sorting
5. Mean calculation
6. Median calculation
7. Mode calculation
8. Standard deviation
9. Minimum and maximum analysis
10. Range calculation
11. Grouping
12. Aggregation
13. Correlation analysis
14. Heatmap visualization
15. Scatter plot analysis
16. Bar chart analysis
17. Box plot analysis
18. Geographical analysis
19. House price categorization
20. Cross-tabulation
""")


# ============================================================
# 46. EXPERIMENT 4 COMPLETED
# ============================================================

print("\n")
print("=" * 60)
print("EXPERIMENT 4 COMPLETED SUCCESSFULLY!")
print("=" * 60)
```
