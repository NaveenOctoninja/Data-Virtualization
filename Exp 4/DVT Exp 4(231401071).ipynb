```python
# ============================================================
# EX NO: 4b
# EXPLORATORY DATA ANALYSIS
# DATA INSPECTION, FILTERING, STATISTICAL ANALYSIS
# AND CORRELATION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os


# ============================================================
# LOAD DATASET
# ============================================================

path = "/content"

print("Dataset Path:")
print(path)

csv_files = [f for f in os.listdir(path) if f.endswith(".csv")]

print("\nCSV Files Found:")
print(csv_files)

csv_file = "housing prices.csv"

print("\nSelected CSV File:")
print(csv_file)

df = pd.read_csv(
    os.path.join(path, csv_file),
    low_memory=False
)

print("\nDataset Loaded Successfully")


# ============================================================
# PROGRAM 1 - DATA INSPECTION
# ============================================================

print("\nFirst 5 Records")
display(df.head())

print("\nLast 5 Records")
display(df.tail())


# ============================================================
# PROGRAM 2 - SHAPE
# ============================================================

print("\nShape of Dataset")
print(df.shape)

print("\nRows =", df.shape[0])
print("Columns =", df.shape[1])


# ============================================================
# PROGRAM 3 - DATA TYPES
# ============================================================

print("\nData Types")
print(df.dtypes)

print("\nDataset Information")
df.info()


# ============================================================
# PROGRAM 4 - FILTERING
# Houses with Median House Value Greater Than 300000
# ============================================================

high_value = df[df["median_house_value"] > 300000]

print("\nRecords with Median House Value Greater Than 300000")

display(
    high_value[
        [
            "longitude",
            "latitude",
            "median_income",
            "total_rooms",
            "median_house_value",
            "ocean_proximity"
        ]
    ].head(10)
)


# ============================================================
# PROGRAM 5 - FILTER HIGH MEDIAN INCOME
# ============================================================

high_income = df[df["median_income"] > 5]

print("\nRecords with Median Income Greater Than 5")

display(
    high_income[
        [
            "longitude",
            "latitude",
            "median_income",
            "total_rooms",
            "median_house_value",
            "ocean_proximity"
        ]
    ].head(10)
)


# ============================================================
# PROGRAM 6 - SORTING
# Sort by Median House Value
# ============================================================

sorted_df = df.sort_values(
    by="median_house_value",
    ascending=False
)

print("\nTop Records by Median House Value")

display(
    sorted_df[
        [
            "longitude",
            "latitude",
            "median_income",
            "total_rooms",
            "median_house_value",
            "ocean_proximity"
        ]
    ].head(10)
)


# ============================================================
# PROGRAM 7 - BAR CHART
# TOP 10 RECORDS BY HOUSE VALUE
# ============================================================

top10 = sorted_df.head(10)

plt.figure(figsize=(10, 5))

plt.bar(
    range(len(top10)),
    top10["median_house_value"]
)

plt.xticks(
    range(len(top10)),
    top10["ocean_proximity"],
    rotation=45
)

plt.xlabel("Ocean Proximity")
plt.ylabel("Median House Value")

plt.title("Top 10 Records by Median House Value")

plt.show()


# ============================================================
# PROGRAM 8 - STATISTICAL ANALYSIS
# ============================================================

print("\nMean Median House Value =")
print(df["median_house_value"].mean())

print("\nMedian Median House Value =")
print(df["median_house_value"].median())

print("\nMode Median House Value =")
print(df["median_house_value"].mode())

print("\nStandard Deviation =")
print(df["median_house_value"].std())

print("\nMinimum Median House Value =")
print(df["median_house_value"].min())

print("\nMaximum Median House Value =")
print(df["median_house_value"].max())


# ============================================================
# PROGRAM 9 - DESCRIPTIVE STATISTICS
# ============================================================

print("\nDescriptive Statistics")

print(
    df[
        [
            "total_rooms",
            "total_bedrooms",
            "population",
            "households",
            "median_income",
            "median_house_value"
        ]
    ].describe()
)


# ============================================================
# PROGRAM 10 - GROUPING
# Average House Value by Ocean Proximity
# ============================================================

group = (
    df.groupby("ocean_proximity")["median_house_value"]
    .mean()
    .sort_values(ascending=False)
)

print("\nAverage House Value by Ocean Proximity")

print(group)


# ============================================================
# VISUALIZATION
# ============================================================

group.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.xlabel("Ocean Proximity")
plt.ylabel("Average House Value")

plt.title("Average House Value by Ocean Proximity")

plt.xticks(rotation=45)

plt.show()


# ============================================================
# PROGRAM 11 - AGGREGATION
# ============================================================

summary = df.groupby("ocean_proximity").agg({

    "median_house_value": "mean",

    "median_income": "mean",

    "total_rooms": "mean",

    "total_bedrooms": "mean",

    "population": "mean",

    "households": "mean"

})


print("\nOcean Proximity-wise Aggregation")

print(summary)


# ============================================================
# PROGRAM 12 - CORRELATION
# ============================================================

numeric = df.select_dtypes(
    include=np.number
)

corr = numeric.corr()

print("\nCorrelation Matrix")

print(corr)


# ============================================================
# PROGRAM 13 - CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(12, 8))

sns.heatmap(
    corr,
    annot=False,
    cmap="coolwarm"
)

plt.title(
    "California Housing Dataset Correlation Heatmap"
)

plt.show()


# ============================================================
# PROGRAM 14 - SCATTER PLOT
# Median Income vs Median House Value
# ============================================================

plt.figure(figsize=(7, 5))

plt.scatter(
    df["median_income"],
    df["median_house_value"]
)

plt.xlabel("Median Income")

plt.ylabel("Median House Value")

plt.title(
    "Median Income vs Median House Value"
)

plt.show()


# ============================================================
# PROGRAM 15 - HISTOGRAM
# Median House Value Distribution
# ============================================================

plt.figure(figsize=(7, 5))

plt.hist(
    df["median_house_value"].dropna(),
    bins=30
)

plt.xlabel("Median House Value")

plt.ylabel("Frequency")

plt.title(
    "Distribution of Median House Values"
)

plt.show()


# ============================================================
# PROGRAM 16 - BOXPLOT
# ============================================================

plt.figure(figsize=(7, 5))

sns.boxplot(
    x=df["median_house_value"].dropna()
)

plt.title(
    "Median House Value Boxplot"
)

plt.xlabel("Median House Value")

plt.show()


# ============================================================
# PROGRAM 17 - VALUE COUNTS
# ============================================================

print("\nOcean Proximity Counts")

print(
    df["ocean_proximity"].value_counts()
)


# ============================================================
# PROGRAM 18 - UNIQUE VALUES
# ============================================================

print("\nUnique Ocean Proximity Values")

print(
    df["ocean_proximity"].unique()
)


# ============================================================
# PROGRAM 19 - TOP 10 RECORDS BY MEDIAN INCOME
# ============================================================

print("\nTop 10 Records by Median Income")

top_income = (
    df.sort_values(
        "median_income",
        ascending=False
    )
    .head(10)
)

display(
    top_income[
        [
            "longitude",
            "latitude",
            "median_income",
            "median_house_value",
            "ocean_proximity"
        ]
    ]
)


# ============================================================
# PROGRAM 20 - TOP 10 RECORDS BY TOTAL ROOMS
# ============================================================

print("\nTop 10 Records by Total Rooms")

top_rooms = (
    df.sort_values(
        "total_rooms",
        ascending=False
    )
    .head(10)
)

display(
    top_rooms[
        [
            "longitude",
            "latitude",
            "total_rooms",
            "total_bedrooms",
            "median_house_value",
            "ocean_proximity"
        ]
    ]
)


# ============================================================
# PROGRAM 21 - HOUSE VALUE BY OCEAN PROXIMITY
# ============================================================

print("\nHouse Value by Ocean Proximity")

ocean_value = (
    df.groupby("ocean_proximity")
    ["median_house_value"]
    .sum()
    .sort_values(ascending=False)
)

print(ocean_value)


# ============================================================
# PROGRAM 22 - AVERAGE INCOME BY OCEAN PROXIMITY
# ============================================================

print("\nAverage Median Income by Ocean Proximity")

income_ocean = (
    df.groupby("ocean_proximity")
    ["median_income"]
    .mean()
    .sort_values(ascending=False)
)

print(income_ocean)


# ============================================================
# PROGRAM 23 - AVERAGE ROOMS BY OCEAN PROXIMITY
# ============================================================

print("\nAverage Total Rooms by Ocean Proximity")

rooms_ocean = (
    df.groupby("ocean_proximity")
    ["total_rooms"]
    .mean()
    .sort_values(ascending=False)
)

print(rooms_ocean)


# ============================================================
# PROGRAM 24 - GEOGRAPHICAL SCATTER PLOT
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    df["longitude"],
    df["latitude"],
    s=5,
    alpha=0.5
)

plt.xlabel("Longitude")

plt.ylabel("Latitude")

plt.title(
    "Geographical Distribution of Housing Records"
)

plt.show()


# ============================================================
# PROGRAM 25 - INCOME AND HOUSE VALUE SCATTER PLOT
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    df["median_income"],
    df["median_house_value"],
    alpha=0.4
)

plt.xlabel("Median Income")

plt.ylabel("Median House Value")

plt.title(
    "Relationship Between Median Income and House Value"
)

plt.show()


# ============================================================
# PROGRAM 26 - POPULATION VS HOUSE VALUE
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    df["population"],
    df["median_house_value"],
    alpha=0.4
)

plt.xlabel("Population")

plt.ylabel("Median House Value")

plt.title(
    "Population vs Median House Value"
)

plt.show()


# ============================================================
# PROGRAM 27 - TOP OCEAN PROXIMITY CATEGORY
# ============================================================

print("\nMost Common Ocean Proximity Category")

print(
    df["ocean_proximity"]
    .value_counts()
    .head(1)
)


# ============================================================
# PROGRAM 28 - FINAL STATISTICAL REPORT
# ============================================================

print("\n==============================================")
print("CALIFORNIA HOUSING EDA REPORT")
print("==============================================")

print(
    "Dataset Shape :",
    df.shape
)

print(
    "Total Records :",
    len(df)
)

print(
    "Number of Ocean Proximity Categories :",
    df["ocean_proximity"].nunique()
)

print(
    "Average Median House Value :",
    df["median_house_value"].mean()
)

print(
    "Minimum Median House Value :",
    df["median_house_value"].min()
)

print(
    "Maximum Median House Value :",
    df["median_house_value"].max()
)

print(
    "Average Median Income :",
    df["median_income"].mean()
)

print(
    "Average Total Rooms :",
    df["total_rooms"].mean()
)

print(
    "Average Population :",
    df["population"].mean()
)

print(
    "Correlation Matrix Generated"
)

print(
    "All EDA Visualizations Generated"
)

print("==============================================")


# ============================================================
# END OF EXPERIMENT
# ============================================================

print("\nEXPERIMENT COMPLETED SUCCESSFULLY")
```
