```python
import pandas as pd
import matplotlib.pyplot as plt

# Load California Housing Dataset
df = pd.read_csv(
    "/content/housing prices.csv",
    low_memory=False
)

# --------------------------------
# 1. Line Chart – House Value by Ocean Proximity
# --------------------------------

house_value = (
    df.groupby("ocean_proximity")["median_house_value"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

plt.plot(
    house_value.index,
    house_value.values,
    marker="o"
)

plt.title("Average House Value by Ocean Proximity")
plt.xlabel("Ocean Proximity")
plt.ylabel("Average Median House Value")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()


# --------------------------------
# 2. Bar Chart – Houses by Ocean Proximity
# --------------------------------

house_count = (
    df["ocean_proximity"]
    .value_counts()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

plt.bar(
    house_count.index,
    house_count.values
)

plt.title("Number of Houses by Ocean Proximity")
plt.xlabel("Ocean Proximity")
plt.ylabel("Number of Houses")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()


# --------------------------------
# 3. Pie Chart – Housing Distribution
# --------------------------------

housing_distribution = (
    df["ocean_proximity"]
    .value_counts()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 8))

plt.pie(
    housing_distribution.values,
    labels=housing_distribution.index,
    autopct="%1.1f%%"
)

plt.title("Housing Distribution by Ocean Proximity")

plt.show()
```
