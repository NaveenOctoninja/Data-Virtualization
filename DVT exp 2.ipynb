```python
# ============================================================
# EXPERIMENT 2
# California Housing Price Analysis and Prediction
# CRISP-DM, SEMMA AND KDD METHODOLOGIES
# ============================================================


# ============================================================
# 1. IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 2. LOAD THE DATASET
# ============================================================

# Change the filename if your CSV has a different name

df = pd.read_csv("/content/housing prices.csv")

print("Dataset Loaded Successfully!")

print("\nFirst 5 Rows:")
print(df.head())


# ============================================================
# 3. DATASET INFORMATION
# ============================================================

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()


# ============================================================
# 4. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# 5. CHECK DUPLICATES
# ============================================================

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())


# ============================================================
# 6. DESCRIPTIVE STATISTICS
# ============================================================

print("\nDescriptive Statistics:")
print(df.describe())


# ============================================================
# ============================================================
#                  CRISP-DM METHODOLOGY
# ============================================================
# ============================================================


# ============================================================
# CRISP-DM STEP 1: BUSINESS UNDERSTANDING
# ============================================================

print("\n")
print("=" * 60)
print("CRISP-DM STEP 1: BUSINESS UNDERSTANDING")
print("=" * 60)

print("""
Objective:
Predict the median house value of a California housing
district using demographic, geographic and housing-related
features.

Target Variable:
median_house_value

The prediction problem is a REGRESSION problem because
the target variable is a continuous numerical value.
""")


# ============================================================
# CRISP-DM STEP 2: DATA UNDERSTANDING
# ============================================================

print("\n")
print("=" * 60)
print("CRISP-DM STEP 2: DATA UNDERSTANDING")
print("=" * 60)

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
print(df.describe())


# ============================================================
# CRISP-DM STEP 3: DATA PREPARATION
# ============================================================

print("\n")
print("=" * 60)
print("CRISP-DM STEP 3: DATA PREPARATION")
print("=" * 60)


# Separate features and target

X = df.drop(
    "median_house_value",
    axis=1
)

y = df["median_house_value"]


# Numerical columns

numeric_features = [
    "longitude",
    "latitude",
    "housing_median_age",
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income"
]


# Categorical columns

categorical_features = [
    "ocean_proximity"
]


print("Numerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)


# ============================================================
# NUMERICAL DATA PREPROCESSING
# ============================================================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        )
    ]
)


# ============================================================
# CATEGORICAL DATA PREPROCESSING
# ============================================================

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),

        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)


# ============================================================
# COMBINE PREPROCESSING
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),

        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)


# ============================================================
# TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Data Shape:")
print(X_train.shape)

print("\nTesting Data Shape:")
print(X_test.shape)


# ============================================================
# CRISP-DM STEP 4: MODELING
# ============================================================

print("\n")
print("=" * 60)
print("CRISP-DM STEP 4: MODELING")
print("=" * 60)


# Random Forest Regression Model

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)


# Create complete pipeline

crisp_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),

        (
            "regressor",
            rf_model
        )
    ]
)


# Train model

crisp_model.fit(
    X_train,
    y_train
)

print("Random Forest Regression Model Trained Successfully!")


# ============================================================
# CRISP-DM STEP 5: EVALUATION
# ============================================================

print("\n")
print("=" * 60)
print("CRISP-DM STEP 5: EVALUATION")
print("=" * 60)


# Prediction

y_pred = crisp_model.predict(
    X_test
)


# Evaluation metrics

crisp_mae = mean_absolute_error(
    y_test,
    y_pred
)

crisp_mse = mean_squared_error(
    y_test,
    y_pred
)

crisp_rmse = np.sqrt(
    crisp_mse
)

crisp_r2 = r2_score(
    y_test,
    y_pred
)


print("Mean Absolute Error (MAE):")
print(crisp_mae)

print("\nMean Squared Error (MSE):")
print(crisp_mse)

print("\nRoot Mean Squared Error (RMSE):")
print(crisp_rmse)

print("\nR2 Score:")
print(crisp_r2)


# ============================================================
# CRISP-DM ACTUAL VS PREDICTED
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.5
)

plt.xlabel(
    "Actual House Value"
)

plt.ylabel(
    "Predicted House Value"
)

plt.title(
    "CRISP-DM: Actual vs Predicted House Value"
)

plt.show()


# ============================================================
# ============================================================
#                     SEMMA METHODOLOGY
# ============================================================
# ============================================================


# ============================================================
# SEMMA STEP 1: SAMPLE
# ============================================================

print("\n")
print("=" * 60)
print("SEMMA STEP 1: SAMPLE")
print("=" * 60)


# Create a random sample of the dataset

sample_df = df.sample(
    frac=1.0,
    random_state=42
).reset_index(drop=True)


print("Original Dataset Shape:")
print(df.shape)

print("\nSample Dataset Shape:")
print(sample_df.shape)

print("\nSample Data:")
print(sample_df.head())


# ============================================================
# SEMMA STEP 2: EXPLORE
# ============================================================

print("\n")
print("=" * 60)
print("SEMMA STEP 2: EXPLORE")
print("=" * 60)


print("Dataset Information:")
sample_df.info()


print("\nStatistical Summary:")
print(sample_df.describe())


print("\nMissing Values:")
print(sample_df.isnull().sum())


# Correlation

sample_corr = sample_df.corr(
    numeric_only=True
)

print("\nCorrelation with Median House Value:")

print(
    sample_corr[
        "median_house_value"
    ].sort_values(
        ascending=False
    )
)


# ============================================================
# SEMMA EXPLORATION VISUALIZATION
# ============================================================

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=sample_df,
    x="median_income",
    y="median_house_value",
    alpha=0.4
)

plt.title(
    "SEMMA Explore: Median Income vs House Value"
)

plt.xlabel(
    "Median Income"
)

plt.ylabel(
    "Median House Value"
)

plt.show()


# ============================================================
# SEMMA STEP 3: MODIFY
# ============================================================

print("\n")
print("=" * 60)
print("SEMMA STEP 3: MODIFY")
print("=" * 60)


# Separate features and target

X_semma = sample_df.drop(
    "median_house_value",
    axis=1
)

y_semma = sample_df[
    "median_house_value"
]


# Train test split

X_train_semma, X_test_semma, y_train_semma, y_test_semma = train_test_split(
    X_semma,
    y_semma,
    test_size=0.20,
    random_state=42
)


print("Training Data:")
print(X_train_semma.shape)

print("\nTesting Data:")
print(X_test_semma.shape)


# ============================================================
# SEMMA STEP 4: MODEL
# ============================================================

print("\n")
print("=" * 60)
print("SEMMA STEP 4: MODEL")
print("=" * 60)


semma_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),

        (
            "regressor",
            RandomForestRegressor(
                n_estimators=100,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)


# Train SEMMA model

semma_model.fit(
    X_train_semma,
    y_train_semma
)

print(
    "SEMMA Random Forest Regression Model "
    "Trained Successfully!"
)


# ============================================================
# SEMMA STEP 5: ASSESS
# ============================================================

print("\n")
print("=" * 60)
print("SEMMA STEP 5: ASSESS")
print("=" * 60)


# Prediction

semma_pred = semma_model.predict(
    X_test_semma
)


# Metrics

semma_mae = mean_absolute_error(
    y_test_semma,
    semma_pred
)

semma_mse = mean_squared_error(
    y_test_semma,
    semma_pred
)

semma_rmse = np.sqrt(
    semma_mse
)

semma_r2 = r2_score(
    y_test_semma,
    semma_pred
)


print("Mean Absolute Error (MAE):")
print(semma_mae)

print("\nMean Squared Error (MSE):")
print(semma_mse)

print("\nRoot Mean Squared Error (RMSE):")
print(semma_rmse)

print("\nR2 Score:")
print(semma_r2)


# ============================================================
# SEMMA ACTUAL VS PREDICTED
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test_semma,
    semma_pred,
    alpha=0.5
)

plt.xlabel(
    "Actual House Value"
)

plt.ylabel(
    "Predicted House Value"
)

plt.title(
    "SEMMA: Actual vs Predicted House Value"
)

plt.show()


# ============================================================
# ============================================================
#                       KDD METHODOLOGY
# ============================================================
# ============================================================


# ============================================================
# KDD STEP 1: SELECTION
# ============================================================

print("\n")
print("=" * 60)
print("KDD STEP 1: SELECTION")
print("=" * 60)


selected_columns = [
    "longitude",
    "latitude",
    "housing_median_age",
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income",
    "ocean_proximity",
    "median_house_value"
]


kdd_df = df[
    selected_columns
].copy()


print("Selected Columns:")
print(kdd_df.columns.tolist())

print("\nSelected Dataset Shape:")
print(kdd_df.shape)


# ============================================================
# KDD STEP 2: PREPROCESSING
# ============================================================

print("\n")
print("=" * 60)
print("KDD STEP 2: PREPROCESSING")
print("=" * 60)


print("Missing Values Before Preprocessing:")

print(
    kdd_df.isnull().sum()
)


# Separate features and target

X_kdd = kdd_df.drop(
    "median_house_value",
    axis=1
)

y_kdd = kdd_df[
    "median_house_value"
]


# Train test split

X_train_kdd, X_test_kdd, y_train_kdd, y_test_kdd = train_test_split(
    X_kdd,
    y_kdd,
    test_size=0.20,
    random_state=42
)


# ============================================================
# KDD STEP 3: TRANSFORMATION
# ============================================================

print("\n")
print("=" * 60)
print("KDD STEP 3: TRANSFORMATION")
print("=" * 60)


# Numerical missing values are replaced by median.
# Categorical missing values are replaced by the mode.
# Categorical values are transformed using One-Hot Encoding.


kdd_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            Pipeline(
                steps=[
                    (
                        "imputer",
                        SimpleImputer(
                            strategy="median"
                        )
                    )
                ]
            ),
            numeric_features
        ),

        (
            "cat",
            Pipeline(
                steps=[
                    (
                        "imputer",
                        SimpleImputer(
                            strategy="most_frequent"
                        )
                    ),

                    (
                        "encoder",
                        OneHotEncoder(
                            handle_unknown="ignore"
                        )
                    )
                ]
            ),
            categorical_features
        )
    ]
)


print(
    "Data transformation pipeline "
    "created successfully!"
)


# ============================================================
# KDD STEP 4: DATA MINING
# ============================================================

print("\n")
print("=" * 60)
print("KDD STEP 4: DATA MINING")
print("=" * 60)


kdd_model = Pipeline(
    steps=[
        (
            "preprocessor",
            kdd_preprocessor
        ),

        (
            "regressor",
            RandomForestRegressor(
                n_estimators=100,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)


# Train KDD model

kdd_model.fit(
    X_train_kdd,
    y_train_kdd
)

print(
    "KDD Random Forest Regression Model "
    "Trained Successfully!"
)


# ============================================================
# KDD STEP 5: INTERPRETATION / EVALUATION
# ============================================================

print("\n")
print("=" * 60)
print("KDD STEP 5: INTERPRETATION / EVALUATION")
print("=" * 60)


# Prediction

kdd_pred = kdd_model.predict(
    X_test_kdd
)


# Evaluation

kdd_mae = mean_absolute_error(
    y_test_kdd,
    kdd_pred
)

kdd_mse = mean_squared_error(
    y_test_kdd,
    kdd_pred
)

kdd_rmse = np.sqrt(
    kdd_mse
)

kdd_r2 = r2_score(
    y_test_kdd,
    kdd_pred
)


print("Mean Absolute Error (MAE):")
print(kdd_mae)

print("\nMean Squared Error (MSE):")
print(kdd_mse)

print("\nRoot Mean Squared Error (RMSE):")
print(kdd_rmse)

print("\nR2 Score:")
print(kdd_r2)


# ============================================================
# KDD ACTUAL VS PREDICTED
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test_kdd,
    kdd_pred,
    alpha=0.5
)

plt.xlabel(
    "Actual House Value"
)

plt.ylabel(
    "Predicted House Value"
)

plt.title(
    "KDD: Actual vs Predicted House Value"
)

plt.show()


# ============================================================
# FINAL COMPARISON
# ============================================================

print("\n")
print("=" * 70)
print("FINAL COMPARISON OF CRISP-DM, SEMMA AND KDD")
print("=" * 70)


comparison = pd.DataFrame({
    "Methodology": [
        "CRISP-DM",
        "SEMMA",
        "KDD"
    ],

    "MAE": [
        crisp_mae,
        semma_mae,
        kdd_mae
    ],

    "MSE": [
        crisp_mse,
        semma_mse,
        kdd_mse
    ],

    "RMSE": [
        crisp_rmse,
        semma_rmse,
        kdd_rmse
    ],

    "R2 Score": [
        crisp_r2,
        semma_r2,
        kdd_r2
    ]
})


print(comparison)


# ============================================================
# VISUAL COMPARISON OF R2 SCORES
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    comparison["Methodology"],
    comparison["R2 Score"]
)

plt.title(
    "R2 Score Comparison"
)

plt.xlabel(
    "Methodology"
)

plt.ylabel(
    "R2 Score"
)

plt.show()


# ============================================================
# EXPERIMENT 2 COMPLETED
# ============================================================

print("\n")
print("=" * 60)
print("EXPERIMENT 2 COMPLETED SUCCESSFULLY!")
print("=" * 60)

print("""
California Housing Price Prediction was performed using:

1. CRISP-DM
2. SEMMA
3. KDD

Random Forest Regression was used as the
data mining / machine learning technique.
""")
```
